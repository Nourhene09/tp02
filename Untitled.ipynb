# Étape 2 — Créer un jeu de données expérimentales

```python
heures = np.array([0, 1, 2, 3, 4, 5])
cellules = np.array([100, 150, 210, 300, 430, 600])
print("Heures :", heures)
print("Cellules :", cellules)
```

# Calculer des valeurs statistiques simples

```python
croissance_moyenne = np.mean(np.diff(cellules))
print("Croissance moyenne (cellules/heure) :", croissance_moyenne)
```

```python
import matplotlib.pyplot as plt
plt.plot(heures, cellules, marker='o', color='green')
plt.title("Croissance cellulaire au fil du temps")
plt.xlabel("Temps (heures)")
plt.ylabel("Nombre de cellules")
```

# EXO 2 — Calculs avec NumPy et création d’un DataFrame Pandas


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
# Exemple de données avec NumPy
data = np.array([10, 20, 30, 40, 50])
mean_val = data.mean()
std_val = data.std()
print("Moyenne :", mean_val)
print("Écart type :", std_val)
# Exemple de DataFrame avec Pandas
df = pd.DataFrame({
'Age': [23, 45, 31, 22, 34, 66],
'Taille': [1.70, 1.85, 1.60, 1.75, 1.80, 1.90]
})
print(df)
print(df.describe())
```

# Tracer un nuage de points (Scatter plot) et un histogramme

```python
# Graphique 1 : Nuage de points (Scatter plot)
plt.figure()
plt.scatter(df['Age'], df['Taille'])
plt.title("Relation Age- Taille")
plt.xlabel("Age")
plt.ylabel("Taille")
plt.grid(True)
plt.show()
# Graphique 2 : Histogramme de distribution
plt.figure()
plt.hist(df['Age'], bins=5)
plt.title("Distribution des ages")
plt.xlabel("Age")
plt.ylabel("Fréquence")
plt.grid(True)
plt.show()
```

# Étape 2 — Lancer le Dashboard démographique (Exemple du cours)

```python
import dash
from dash import dcc, html
import plotly.express as px
import pandas as pd

# Exemple de données
df_villes = pd.DataFrame({
    "Ville": ["Alger", "Oran", "Constantine"],
    "Population": [3500000, 2000000, 1000000]
})

fig = px.bar(df_villes, x="Ville", y="Population", title="Population par ville")

# Création de l'application
app = dash.Dash(__name__)

app.layout = html.Div([
    html.H1("Dashboard démographique"),
    dcc.Graph(
    figure=fig)
])

if __name__ == "__main__":
    app.run(debug=True)
```

```python
import plotly.express as px
import pandas as pd

# Exemple de données
df = pd.DataFrame({
    "Age": [23, 45, 32],
    "Score": [88, 92, 75]
})

fig = px.scatter(df, x="Age", y="Score", title="Age vs score ")
fig.show()
fig = px.histogram(df, x="Score", nbins=5, title="Distribution des scores ")
fig.show()
```
